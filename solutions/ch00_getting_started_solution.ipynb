```python
def double(x):
    """Return twice x."""
    return 2 * x

check("double", double, [(3, 6), (0, 0), (-2, -4)])
```

**Correct** means your function matched every check, but a check can also give you one of three other messages.

- **Not started** means the function still has its placeholder line.
- **Not yet** means it runs but gives the wrong answer, often with a hint about why.
- **Error** means the code raised an error, and you'll see the message.
